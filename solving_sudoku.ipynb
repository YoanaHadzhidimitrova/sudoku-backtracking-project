## Problem Statement

The Sudoku board contains 81 cells that are arranged into 9x9 grid, which have to be filled with numbers. Initially, there are some numbers that are populated already in the board and the remaining numbers should be filled based on some rules and constraints. The numbers that can be used are between 1 and 9 inclusively and must appear exactly once in every column, row and small 3x3 section. We can represent each blank cell as a variable $x_{i,j}$, where $i$ represents row index and $j$ represents column index $$ x_{i,j} \in \{1, 2, 3, 4, 5, 6, 7, 8, 9\} $$

If every cell can independently take number from 1 to 9, the number of possible configurations of the Sudoku board would be $9^{81}$. However, we have to follow the constraints, described above in order to have a valid Sudoku board, so this reduces the possible configurations. We need a systematic search method, that will explore the board and eliminate the invalid options in early stage. Backtracking algorithm is a method for solving problems by incrementally building candidates and abandoning them as soon as they violate some of the rules and constraints.

In this project, Sudoku is formulated as constraint satisfaction problem and we will implement Backtracking algorithm in Python in order to solve it.

## Mathematical Background

### Combinatorics

Combinatorics is a branch of mathematics, that studies different ways of arranging elements and counting the number of possible configurations that satisfy specific conditions

Sudoku is considered a combinatorial problem, because it is puzzle game that in which the numbers from 1 to 9 have to be arranged following several rules. In each row, column, 3x3 section we can have $9!$ different combinations which is 362 880 combinations in each section. When solving Sudoku puzzle, the solver must determine the correct values for empty cells while satisfying all puzzle constraints. However, most of the possible combinations are invalid, because they violate Sudoku's rules. As a result, the number valid solutions is small, but the search space that must be explored is still very large. Due to this, the solver must use systematic search strategy that eliminates invalid possibilities early, such as Backtracking algorithm.

### Constraint Satisfaction Problems

A Constraint Satisfaction Problem (CSP) is a mathematical problem, where the solution must meet different constraints. In CSP the objective is to assign values to variables, which meet specific constraints and lay on specific domains. The three components of CSP are variables, domains and constraints, formally defined as triple $CSP = (X, D, C)$, where:\
$X = {x_1, x_2, ..., x_n}$ is a set of **variables** - the things we need to find values for. In our Sudoku example, variables are all empty cells for which we need to find a suitable number for.\
$D = {D_1, D_2, ..., D_n}$ is a set of **domains** - the specific values that can be assigned for variables. In our Sudoku example, domains are the numbers from 1 to 9 inclusively.\
$C$ is a set of **constraints** - the rules which restrict how the variables can be assigned. In our Sudoku example, the restriction is that in every row, column and 3x3 subgrid we must contain the numbers from 1 to 9 exactly once.

The **backtracking algorithm** is a method to systematicaly explore possible solutions in CSPs. It operates by assigning values to empty cells and backtracking if the value violates any rule. The algorithm works with the following steps:

    1. Finds variable and assigns value
    2. It recursively assigns values to subsequent cells
    3. When the assigned value is not correct, the algorith goes to the previous value and tries with different value.
    4. The process continues either until valid value is assigned or all possible options are tried.
    
The method is widely used due to its simplicity, but it can be inefficient for large sets with many variables.

### Search Space

When solving a Sudoku puzzle, each cell should be assigned a number from 1 to 9, which means that for each cell we can have 9 options as a result. One 9x9 grid contains 81 cells, which means that if we have to assign the numbers independantly without following any rules, we would have $9^{81}$ possible combinations, which represents very large search space. When we create rules and constraints, this large number of possible configurations significantly drops as many of the combinations will be invalid. Because of this large search space, we can use algorithms like backtracking in order to explore the possible variations and eliminate part of them in the early stages.

## Methodology

### Backtracking Algorithm

**Backtracking algorithm** is problem-solving technique that finds solution by trying different possibilities and undoing them if they lead to a dead end. The algorithm is commonly used in puzzle problems and constraint satisfaction problems. When it reaches a dead end, it goes to the previous step and explores new possibilities until valid path is found or all options are exhausted.

Steps which Backtracking algorithm is following, when solving Sudoku puzzle:

    1. Find an empty cell (0)
    2. Attempt digits from 1 to 9
    3. Check row, column and subgrid constraints
    4. If the attempt is not valid, go back to the previous attempt and try different number.
    5. Repeat steps 2-4 until valid solution is found or all possible options are exhausted.

The backtracking algorithm can be viewed as a traversal of a search tree, where each node represents a partial assignment of values. Each recursive call corresponds to moving deeper in the tree, while backtracking corresponds to returning to a previous node when a constraint violation occurs.

### Minimum Remaining Values (MRV) heuristic

Minimum Remaining Values (MRV) is a heuristic that prioritizes selecting variables with the fewest remaining legal moves in a search algorithm, helping to reduce the complexity of the search space. This approach is particularly useful in solving constraint satisfaction problems by efficiently narrowing down possibilities.

### Finding an Empty Cell

The function `find_empty()` searches the Sudoku grid for the first empty cell. Empty cells are represented with the value 0. The function returns the position of the empty cell as a pair of indices (row, column). If no empty cell is found, the function returns `None`, which indicates that the board is already complete.

```python
def find_empty(board):
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                return i, j
            else:
                if i == 8 and j == 8:
                    if board[i][j] != 0:
                        return None
```

```python
empty_cell_indexes = find_empty(board)
```

### Constraint Validation

The function `valid()` checks whether a candidate number can be placed in a specific cell without violating Sudoku constraints. The function verifies 
three conditions:

1. The number does not already appear in the same row.
2. The number does not appear in the same column.
3. The number does not appear in the corresponding 3×3 subgrid.

If all constraints are satisfied, the function returns `True`; otherwise it returns `False`.

```python
def valid(board, number, empty_cell_indexes):
    if number < 1 or number > 9:
        return False
    row = empty_cell_indexes[0]
    column = empty_cell_indexes[1]
    subgrid_start_row = (row // 3)*3
    subgrid_start_col = (column // 3)*3
    for r in range(0, 9):
        if board[row][r] == number:
            return False
    for c in range(0, 9):
        if board[c][column] == number:
            return False
    for sr in range(subgrid_start_row, subgrid_start_row + 3):
        for sc in range(subgrid_start_col, subgrid_start_col + 3):
            if board[sr][sc] == number:
                return False
    return True
```

### Backtracking Solver

The function `solve()` implements the recursive backtracking algorithm used to complete the Sudoku puzzle. The algorithm repeatedly finds an empty cell and attempts to place numbers from 1 to 9 in that position.
For each candidate number, the function checks whether the placement satisfies all Sudoku constraints. If the placement is valid, the algorithm proceeds recursively to solve the remaining board. If no valid number leads to a solution, the algorithm backtracks by resetting the cell and trying another value.

```python
def solve(board):
    global steps
    steps += 1
    row_column = find_empty(board)
    if row_column == None:
        return True

    for i in range(1,10):
        if valid(board, i, row_column) == True:
            row = row_column[0]
            column = row_column[1]
            board[row][column] = i
            if solve(board) == True:
                return True
            else: 
                board[row][column] = 0
            
    return False
```

### Algorithm Complexity

The backtracking algorithm explores the search space of possible Sudoku configurations by recursively assigning values to empty cells. In the worst case, each empty cell could take up to 9 possible values, which leads to a theoretical worst-case complexity of approximately O(9^n), where n is the 
number of empty cells.
However, Sudoku constraints significantly reduce the number of valid candidate values, allowing the algorithm to prune large portions of the search space. In practice, this makes backtracking efficient enough to solve typical Sudoku puzzles within a short time.

```python
print("Original board:")
print_board(board)

print(solve(board))

print("Solved board:")
print_board(board)
```

## Heuristic Optimization of the Search Process

While the standard backtracking algorithm guarantees a correct solution, its efficiency depends on the order in which variables (empty cells) are selected during the search. In its basic form, the algorithm selects the first available empty cell, which may lead to unnecessary exploration of the search space.
From a constraint satisfaction perspective, the performance of the search can be improved by incorporating heuristics that reduce the branching factor. One such approach is the Minimum Remaining Values (MRV) heuristic, which prioritizes variables with the smallest number of feasible values.

## Minimum Remaining Values (MRV)

The Minimum Remaining Values heuristic selects a variable (empty cell) with the smallest domain of valid candidate values. Formally, for each unassigned variable $x_{i,j} $, the size of its domain is defined as the number of the values in ${1,...,9}$ that satisfy all Sudoku constraints.
Instead of choosing the first unassigned variable, the MRV heuristic selects the variable with the minimum domain size:
$$
x^* = \arg\min_{x_{i,j}} |D_{i,j}|
$$
where $D_{i,j}$ represents the set of valid candidates for cell $ i,j $
This strategy reduces the branching factor of the search tree and increases the likelihood of early detection of constraints violations, thereby improving the efficiency of the backtracking algorithm
The MRV heuristic is implemented by evaluating the number of valid candidate values for each empty cell and selecting the most constrained variable at each step of the recursion.

The function `count_valid_options()` counts the number of valid values that can be assigned to a given empty cell. It iterates between the numbers from 1 to 9 and counts the possible solutions.

```python
def count_valid_options(board, empty_cell_indexes):
    row = empty_cell_indexes[0]
    column = empty_cell_indexes[1]
    counter = 0
    for num in range(1, 10):
        if valid(board, num, empty_cell_indexes):
            counter+=1

    return counter
```

The function `find_best_empty()` selects the next empty cell to fill using the Minimum Remaining Values heuristic. It checks which of the cells has mnimum valid values as solutions, reducing the branching factor of the search

```python
def find_best_empty(board):
    best_cell = None
    min_options = 10
    for i in range(9):
        for j in range(9):
            if board[i][j] == 0:
                options = count_valid_options(board, (i, j))

                if options < min_options:
                    min_options = options
                    best_cell = (i, j)

    return best_cell            
```

The function `solve_mrv()` solves the Sudoku puzzle using recursive backtracking enhanced with the MRV heuristic. 
At each step, it selects the most constrained empty cell, tries valid values, and backtracks if a conflict occurs, until a solution is found.

```python
def solve_mrv(board):
    global steps
    steps += 1
    row_column = find_best_empty(board)
    if row_column is None:
        return True

    for i in range(1, 10):
        if valid(board, i, row_column):
            row = row_column[0]
            column = row_column[1]
            board[row][column] = i

            if solve_mrv(board):
                return True
            board[row][column] = 0
    return False
```

## Discussion and Limitations


The Sudoku problem was modeled as a constraint satisfaction problem and solved using recursive backtracking. While the MRV heuristic is designed to improve efficiency by reducing the branching factor, the experimental results reveal an important trade-off.

Although MRV reduced the number of recursive steps, it significantly increased execution time. This is due to the overhead introduced by repeatedly scanning the board and computing valid options for each empty cell. In contrast, the basic backtracking approach, while exploring more states, performs less computation per step.

This highlights a key limitation: heuristics such as MRV do not always guarantee faster performance in practice, especially for smaller problem instances or when the cost of the heuristic itself is high.

More advanced techniques, such as forward checking, constraint propagation, or optimized data structures, could help reduce this overhead and better leverage the advantages of MRV.
